#### Append and Insect

```python
new_arry1 = np.ones((3,3,3))
new_arry1
```

```python
new_arry2 = np.zeros((3,3,3))
new_arry2
```

```python
appd = np.append(new_arry1,new_arry2, axis=0)
appd
```

```python
appd = np.append(new_arry1,new_arry2, axis=1)
appd
```

```python
inserted_arry = np.insert(new_arry2, 2, new_arry1, axis=0)
inserted_arry
```
