## 4. Data Perprocessing

```python
df.shape
```

```python
# splitting the data into training and testing datasets

data_training = pd.DataFrame(df.Close[0:int(len(df)*0.7)]) # 70% of training
data_testing = pd.DataFrame(df.Close[int(len(df)*0.7):int(len(df))]) # 30% of testing

print(data_training)
print(data_testing)
```

```python
data_training
```

```python
# Scaling down the between 0 and 1
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler(feature_range=(0,1))
```

```python
data_training_array = scaler.fit_transform(data_training)
data_training_array
```

## 5. Sequence Creation

```python
x_train = []
y_train = []

for i in range(100,data_training_array.shape[0]):
    x_train.append(data_training_array[i-100:i])
    y_train.append(data_training_array[i,0])

x_train
```

```python
x_train, y_train = np.array(x_train), np.array(y_train)
```

## 6. Model Building

```python
!pip install tensorflow
```

```python
import tensorflow as tf
print(tf.__version__)
```

```python
# ml models
from keras.models import Sequential
from keras.layers import Dense, LSTM, Input
```

```python
#initialize the sequential model
model = Sequential()

model.add(Input(shape=(100,1)))
model.add(LSTM(units=128, activation='tanh', return_sequences=True))
model.add(LSTM(units=64))
model.add(Dense(25))
model.add(Dense(1))
```

## 7. Model Training

```python
model.compile(optimizer='adam', loss= 'mean_squared_error')
model.fit(x_train, y_train, epochs=50)
```

```python
model.summary()
```

```python
model.save('lstm_trained_model.keras')
```

## 8. Preparing Test Data

```python
data_testing
```

```python
past_100_days= data_training.tail(100)
past_100_days
```

```python
final_df = pd.concat([past_100_days, data_testing], ignore_index=True)
final_df
```

```python
input_data = scaler.fit_transform(final_df)
input_data
```

```python
x_test =[]
y_test = []

for i in range(100, input_data.shape[0]):
    x_test.append(input_data[i-100:i])
    y_test.append(input_data[i,0])
```

```python
x_test, y_test = np.array(x_test), np.array(y_test)
```

## 9. Making Predictions

```python
y_predicted = model.predict(x_test)
y_predicted
```

```python
y_test
```

```python
y_predicted = scaler.inverse_transform(y_predicted.reshape(-1,1)).flatten()
y_tests = scaler.inverse_transform(y_test.reshape(-1,1)).flatten()
```

## 10. Model Evaluation

```python
# Mean Square Error
from sklearn.metrics import mean_squared_error, r2_score
```

```python
mse = mean_squared_error(y_tests, y_predicted)
print(f"Mean Squared Error (MSE): {mse}")
```

```python
# Root mean square error
rmse = np.sqrt(mse)
print(f"Root Mean Squared Error (RMSE): {rmse}")
```

```python
#R-Square
r2 = r2_score(y_tests, y_predicted)
print(f"R-Squared: {r2}")
```
